```python
import os
import shutil
from glob import glob

def split_data(source_folder='data/train/preprocessed_images', label_folder='data/train/labels', output_folder='dataset_', train_ratio=0.7, valid_ratio=0.2):
    """Splits the dataset and corresponding labels into training, validation, and testing sets."""
    images = glob(os.path.join(source_folder, '*.jpg'))
    labels = glob(os.path.join(label_folder, '*.txt'))  # Assuming labels are in .txt format
    
    print(f"Found {len(images)} images and {len(labels)} labels.")
    
    train_count = int(len(images) * train_ratio)
    valid_count = int(len(images) * valid_ratio)
    
    train_images = images[:train_count]
    valid_images = images[train_count:train_count + valid_count]
    test_images = images[train_count + valid_count:]
    
    for category, image_list in zip(['train', 'valid', 'test'], [train_images, valid_images, test_images]):
        image_folder = os.path.join(output_folder, category, 'images')
        label_folder_out = os.path.join(output_folder, category, 'labels')
        os.makedirs(image_folder, exist_ok=True)
        os.makedirs(label_folder_out, exist_ok=True)
        
        for img_path in image_list:
            img_basename = os.path.basename(img_path)
            shutil.copy(img_path, os.path.join(image_folder, img_basename))
            
            # Copy corresponding label file
            label_path = os.path.join(label_folder, os.path.splitext(img_basename)[0] + '.txt')
            if os.path.exists(label_path):
                shutil.copy(label_path, os.path.join(label_folder_out, os.path.basename(label_path)))
            else:
                print(f"Warning: No label found for image {img_basename}")

    print("Dataset and labels split completed!")
    create_yaml(output_folder)

def create_yaml(output_folder):
    """Creates the data.yaml file for YOLO training with absolute paths."""
    train_path = os.path.abspath(os.path.join(output_folder, 'train/images'))
    valid_path = os.path.abspath(os.path.join(output_folder, 'valid/images'))
    test_path = os.path.abspath(os.path.join(output_folder, 'test/images'))

    yaml_content = f"""train: {train_path}
val: {valid_path}
test: {test_path}

nc: 1  # Number of classes
names: ['gingivitis']  # Modify this based on your classes
"""

    yaml_path = os.path.join(output_folder, 'data.yaml')
    with open(yaml_path, 'w') as f:
        f.write(yaml_content.strip())

    print(f"✅ data.yaml created at {yaml_path}")

split_data()
```

```python
from ultralytics import YOLO

# Load a YOLOv8 model (you can choose yolov8n.pt, yolov8s.pt, yolov8m.pt, etc.)
model = YOLO('yolov8s.pt')  # Start with 's' version for a good balance

# Train the model
model.train(
    data='C:/Users/jilln/LEARNING RESOURCES 2024-2025/3rd Year - 2nd Semester/Research/Gingi Preprocessing/DeepDent/gingivitis/dataset_/data.yaml',
    epochs=100,           # Adjust as needed
    imgsz=640,            # Image size (can also use 512 or 416)
    batch=8,              # Adjust depending on your GPU/CPU
    name='gingivitis_yolov8',
    verbose=True
)
```

```python
import os
from glob import glob
from ultralytics import YOLO

# Load trained YOLOv8 model
model = YOLO('runs/detect/gingivitis_yolov8/weights/best.pt')

# Class names
class_names = ['redness', 'bleeding', 'swelling', 'none']

# Test image folder
test_folder = 'C:/Users/jilln/LEARNING RESOURCES 2024-2025/3rd Year - 2nd Semester/Research/Gingi Preprocessing/DeepDent/gingivitis/dataset_/test/images'
image_paths = glob(os.path.join(test_folder, '*.jpg'))

# Process each image
for image_path in image_paths:
    result_list = model(image_path)  # returns a list of one result per image
    result = result_list[0]          # get the actual result object

    # Safeguard: if no boxes are detected
    if result.boxes is not None and result.boxes.cls is not None:
        predicted_classes = result.boxes.cls.cpu().numpy().astype(int)
        predicted_labels = [class_names[i] for i in predicted_classes]
    else:
        predicted_labels = []

    gingivitis_detected = any(label in ['redness', 'bleeding', 'swelling'] for label in predicted_labels)
    status = '🟥 Gingivitis' if gingivitis_detected else '✅ No Gingivitis'

    print(f"{os.path.basename(image_path)}: {status} (Detected: {predicted_labels})")
```

```python
import os
from glob import glob
from ultralytics import YOLO

# Load trained YOLOv8 model
model = YOLO('runs/detect/gingivitis_yolov8/weights/best.pt')

# Class names (must match the model training order)
class_names = ['redness', 'bleeding', 'swelling', 'none']

# Required symptoms for gingivitis diagnosis
required_symptoms = {'redness', 'bleeding', 'swelling'}

# Test image folder
test_folder = 'C:/Users/jilln/LEARNING RESOURCES 2024-2025/3rd Year - 2nd Semester/Research/Gingi Preprocessing/DeepDent/gingivitis/dataset_/test/images'
image_paths = glob(os.path.join(test_folder, '*.jpg'))

# Process each image
for image_path in image_paths:
    result_list = model(image_path)
    result = result_list[0]

    if result.boxes is not None and result.boxes.cls is not None:
        predicted_classes = result.boxes.cls.cpu().numpy().astype(int)
        predicted_labels = [class_names[i] for i in predicted_classes]
        detected_symptoms = set(predicted_labels)
    else:
        predicted_labels = []
        detected_symptoms = set()

    # Check if ALL required symptoms are present
    gingivitis_detected = required_symptoms.issubset(detected_symptoms)
    status = '🟥 Gingivitis' if gingivitis_detected else '✅ No Gingivitis'

    print(f"{os.path.basename(image_path)}: {status} (Detected: {sorted(detected_symptoms)})")
```
